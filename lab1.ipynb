```python
def task_8():
    raw_input_data = input().strip()
    tokens = raw_input_data.split()

    if not tokens:
        return "Помилка. Введено порожній рядок"

    # перевіряємо, чи всі введені елементи є числами
    arr = []
    for token in tokens:
        try:
            val = int(token)
            arr.append(val)
        except ValueError:
            return "Помилка. У списку присутнє нечислове значення"

    # gnome sort від найбільшого до найменшого
    index = 0
    n = len(arr)
    while index < n:
        if index == 0 or arr[index] <= arr[index - 1]:
            index += 1
        else:
            arr[index], arr[index - 1] = arr[index - 1], arr[index]
            index -= 1

    return arr
```
